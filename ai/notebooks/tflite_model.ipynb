```python
import numpy as np
import tensorflow as tf
from PIL import Image
from pathlib import Path


# ============================================================
# LABELS
# ============================================================

CROP_NAMES = [
    'carrot', 'cucumber', 'mango', 'okra',
    'orange', 'pepper', 'plantain', 'potato', 'tomato'
]

FRESH_NAMES = [
    'aging', 'fresh', 'spoiled'
]


# ============================================================
# CONFIGURATION
# ============================================================

IMG_SIZE = (224, 224)

TFLITE_PATH = "agriconnect.tflite"

TEST_PATH = Path(
    '/Users/addosakah/Desktop/COE3/Sem 2/Software engineering/'
    'AgriConnect project/AgriConnect AI/'
    'AgriConnect_Final_dataset/test/'
)



# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def load_and_preprocess(image_path):
    """
    Loads a real image and prepares it for the TFLite model.

    Resize with padding:
        224 x 224

    Format:
        RGB
        float32
    """

    img = Image.open(image_path).convert("RGB")

    img = np.array(img).astype(np.float32)

    img_tensor = tf.convert_to_tensor(img)

    img_tensor = tf.image.resize_with_pad(
        img_tensor,
        IMG_SIZE[0],
        IMG_SIZE[1]
    )

    img_tensor = tf.expand_dims(
        img_tensor,
        axis=0
    )

    return img_tensor.numpy()


# ============================================================
# LOAD TFLITE MODEL ONCE
# ============================================================

interpreter = tf.lite.Interpreter(
    model_path=TFLITE_PATH
)

interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()


# ============================================================
# PREDICT
# ============================================================

def predict(image_path):

    image_array = load_and_preprocess(image_path)

    # Send image to model
    interpreter.set_tensor(
        input_details[0]['index'],
        image_array
    )

    # Run inference
    interpreter.invoke()

    # Get model outputs
    crop_out = interpreter.get_tensor(
        output_details[0]['index']
    )

    fresh_out = interpreter.get_tensor(
        output_details[1]['index']
    )

    # --------------------------------------------------------
    # Crop prediction
    # --------------------------------------------------------

    crop_index = np.argmax(crop_out[0])

    predicted_crop = CROP_NAMES[crop_index]

    crop_confidence = float(
        crop_out[0][crop_index]
    )


    # --------------------------------------------------------
    # Freshness prediction
    # --------------------------------------------------------

    freshness_index = np.argmax(fresh_out[0])

    predicted_freshness = FRESH_NAMES[
        freshness_index
    ]

    freshness_confidence = float(
        fresh_out[0][freshness_index]
    )


    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print(f"\nImage: {image_path}")

    print(
        f"  Crop      → {predicted_crop} "
        f"(confidence {crop_confidence * 100:.1f}%)"
    )

    print(
        f"  Freshness → {predicted_freshness} "
        f"(confidence {freshness_confidence * 100:.1f}%)"
    )


    # --------------------------------------------------------
    # Top-3 crop predictions
    # --------------------------------------------------------

    top3_idx = np.argsort(
        crop_out[0]
    )[::-1][:3]

    print("  Top-3 crop guesses:")

    for i in top3_idx:

        print(
            f"    {CROP_NAMES[i]}: "
            f"{crop_out[0][i] * 100:.1f}%"
        )


    # --------------------------------------------------------
    # RETURN VALUES
    # --------------------------------------------------------

    return (
        predicted_crop,
        predicted_freshness,
        crop_confidence,
        freshness_confidence
    )


# # ============================================================
# # TEST
# # ============================================================
#
# if __name__ == "__main__":
#
#     test_files = get_test_files()
#
#     print(
#         f"Found {len(test_files)} test images."
#     )
#
#     for item in test_files:
#
#         predict(item["path"])
```

## TEST MODEL

```python
from pathlib import Path
import random


def get_test_files():
    """
    Select exactly ONE random image from EACH valid folder.

    Expected structure:

    test/
        carrot_fresh/
        carrot_aging/
        carrot_spoiled/
        cucumber_fresh/
        ...
    """

    test_files = []

    TEST_PATH = Path(
        '/Users/addosakah/Desktop/COE3/Sem 2/Software engineering/'
        'AgriConnect project/AgriConnect AI/'
        'AgriConnect_Final_dataset/test/'
    )

    for folder in TEST_PATH.iterdir():

        if not folder.is_dir():
            continue

        # Example: "mango_fresh"
        folder_name = folder.name.lower()

        try:
            crop, freshness = folder_name.split("_", 1)
        except ValueError:
            print(f"Skipping invalid folder: {folder.name}")
            continue

        # Validate crop
        if crop not in CROP_NAMES:
            print(f"Unknown crop: {crop}")
            continue

        # Validate freshness
        if freshness not in FRESH_NAMES:
            print(f"Unknown freshness state: {freshness}")
            continue

        # Get images ONLY from this folder
        images = [
            image_path
            for image_path in folder.iterdir()
            if image_path.is_file()
            and image_path.suffix.lower() in {
                ".jpg", ".jpeg", ".png", ".webp"
            }
        ]

        if not images:
            print(f"No images found in {folder.name}")
            continue

        # ONE random image from THIS folder
        selected_image = random.choice(images)

        test_files.append({
            "path": selected_image,
            "actual_crop": crop,
            "actual_freshness": freshness
        })

    return test_files
```

```python
# ============================================================
# EVALUATE TEST DATA
# ============================================================

test_files = get_test_files()

actual_crops = []
predicted_crops = []

actual_freshness = []
predicted_freshness = []

crop_correct = []
freshness_correct = []

results = []


for item in test_files:

    image_path = item["path"]

    actual_crop = item["actual_crop"]
    actual_freshness_label = item["actual_freshness"]

    (
        predicted_crop,
        predicted_freshness_label,
        crop_confidence,
        freshness_confidence
    ) = predict(image_path)

    # Store labels
    actual_crops.append(actual_crop)
    predicted_crops.append(predicted_crop)

    actual_freshness.append(actual_freshness_label)
    predicted_freshness.append(predicted_freshness_label)

    # Correct / wrong
    crop_is_correct = (
        actual_crop == predicted_crop
    )

    freshness_is_correct = (
        actual_freshness_label == predicted_freshness_label
    )

    crop_correct.append(crop_is_correct)
    freshness_correct.append(freshness_is_correct)

    # Store complete result
    results.append({
        "image": image_path.name,
        "path": str(image_path),

        "actual_crop": actual_crop,
        "predicted_crop": predicted_crop,
        "crop_confidence": crop_confidence,
        "crop_correct": crop_is_correct,

        "actual_freshness": actual_freshness_label,
        "predicted_freshness": predicted_freshness_label,
        "freshness_confidence": freshness_confidence,
        "freshness_correct": freshness_is_correct
    })


# ============================================================
# ACCURACY
# ============================================================

crop_correct_count = sum(crop_correct)
crop_wrong_count = len(crop_correct) - crop_correct_count

fresh_correct_count = sum(freshness_correct)
fresh_wrong_count = len(freshness_correct) - fresh_correct_count

crop_accuracy = (
    crop_correct_count / len(crop_correct)
) * 100

freshness_accuracy = (
    fresh_correct_count / len(freshness_correct)
) * 100


print("\n==========================================")
print("AGRICONNECT TEST RESULTS")
print("==========================================")

print(f"Total images:              {len(test_files)}")

print(f"\nCrop classification:")
print(f"  Correct:                  {crop_correct_count}")
print(f"  Wrong:                    {crop_wrong_count}")
print(f"  Accuracy:                 {crop_accuracy:.2f}%")

print(f"\nFreshness classification:")
print(f"  Correct:                  {fresh_correct_count}")
print(f"  Wrong:                    {fresh_wrong_count}")
print(f"  Accuracy:                 {freshness_accuracy:.2f}%")
```

```python
# ============================================================
# CORRECT VS WRONG
# ============================================================
import matplotlib.pyplot as plt

labels = [
    "Crop\nCorrect",
    "Crop\nWrong",
    "Freshness\nCorrect",
    "Freshness\nWrong"
]

values = [
    crop_correct_count,
    crop_wrong_count,
    fresh_correct_count,
    fresh_wrong_count
]

plt.figure(figsize=(10, 6))

bars = plt.bar(labels, values)

plt.title(
    "AgriConnect Model — Correct vs Wrong Predictions"
)

plt.ylabel("Number of Images")

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(int(height)),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CROP CONFUSION MATRIX
# ============================================================
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_crop = confusion_matrix(
    actual_crops,
    predicted_crops,
    labels=CROP_NAMES
)

fig, ax = plt.subplots(figsize=(10, 8))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_crop,
    display_labels=CROP_NAMES
)

disp.plot(
    ax=ax,
    xticks_rotation=45,
    values_format="d"
)

ax.set_title(
    f"Crop Classification\nAccuracy: {crop_accuracy:.2f}%"
)

plt.tight_layout()
plt.show()


# ============================================================
# FRESHNESS CONFUSION MATRIX
# ============================================================

cm_freshness = confusion_matrix(
    actual_freshness,
    predicted_freshness,
    labels=FRESH_NAMES
)

fig, ax = plt.subplots(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_freshness,
    display_labels=FRESH_NAMES
)

disp.plot(
    ax=ax,
    values_format="d"
)

ax.set_title(
    f"Freshness Classification\n"
    f"Accuracy: {freshness_accuracy:.2f}%"
)

plt.tight_layout()
plt.show()
```
